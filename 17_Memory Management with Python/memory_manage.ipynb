#### Reference Counting
Reference counting is the primary method Python uses to manage memory. Each object in python maintains a count of references pointing to it. When the reference count drops to zero, the memory occupied by the object is deallocated.

```python
import sys

a = []
## 2  (one reference from 'a' and one from getrefcount())
print(sys.getrefcount(a))
```

```python
b = a
print(sys.getrefcount(b))
```

```python
del b

print(sys.getrefcount(b))
```

```python
print(sys.getrefcount(a))
```

#### Garbage Collection
Pythoon includes a cyclic garbage collector to handle reference cycles. Reference cycles occur when objects reference each other, preventing their reference counts from reaching zero.

```python
import gc

# enable garbage collection
gc.enable()
```

```python
# disable garbage collection
gc.disable()
```

```python
# manually trigger
gc.collect()
```

```python
## Get garbage collection stats

gc.get_stats()
```

```python
## get unreachable objects

print(gc.garbage)
```

# Memory Management Best Practices

1. **Use Local Variables**
   * Local variables have a shorter lifespan.
   * They clear from memory faster than global variables.

2. **Avoid Circular References**
   * Reference loops prevent proper garbage collection.
   * This can lead to severe memory leaks.

3. **Use Generators**
   * Generators produce items one at a time.
   * They keep your overall memory footprint low.

4. **Explicitly Delete Objects**
   * Use the `del` statement to remove references.
   * This manually clears variables and frees up resources.

5. **Profile Memory Usage**
   * Use tools like `tracemalloc` and `memory_profiler`.
   * These find hidden leaks and optimize your code.


```python
# handling circular reference
import gc

class MyyObject:
    def __init__(self, name):
        self.name = name
        print(f"Object {self.name} created")
    def __del__(self):
        print(f"Object {self.name} deleted ...")


# Create circular reference

obj1 = MyyObject('obj1')
obj2 = MyyObject('obj2')

obj1.ref = obj2
obj2.ref = obj1


del obj1
del obj2


## manually trigegr the garbage collector
gc.collect()
```

```python
# print collected garbage
gc.garbage
```

```python
## Generators for memory efficiency
# Generator allow you to produce one item at a time, using memory efficiency by only keeping one item in memory at a time.

def generate_numbers(n):
    for i in range(n):
        yield i


# using the generator
for num in generate_numbers(100000):
    print(num)

    if num > 10: 
        break
```

```python
# profiling memory usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapsot = tracemalloc.take_snapshot()
    top_stats = snapsot.statistics('lineno')

    print('[ Top 10 ]')
    for stat in top_stats[:10]:
        print(stat)
```

```python
main()
```
