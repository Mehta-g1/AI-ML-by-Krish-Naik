## Support Vector Regression Implimentation

```python
## Dataset (Tips Dataset)

import seaborn as sns

df = sns.load_dataset('tips')
df.head()
```

```python
## Independent and Dependent features

X = df.drop('total_bill', axis=1)
y = df['total_bill']
```

```python
# Train test split

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=10)
```

```python
X_train.sample(5)
```

```python
## Feature Encoding
# Label Encoding
# One hot encoding

from sklearn.preprocessing import LabelEncoder

le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
X_train['sex'] = le1.fit_transform(X_train['sex'])
X_train['smoker'] = le2.fit_transform(X_train['smoker'])
X_train['time'] = le3.fit_transform(X_train['time'])

X_train.sample(10)
```

```python
X_test['sex'] = le1.transform(X_test['sex'])
X_test['smoker'] = le2.transform(X_test['smoker'])
X_test['time'] = le3.transform(X_test['time'])

X_test.sample(10)
```

```python
## Onehot Encoding --- ColumnTransformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct = ColumnTransformer(transformers=[('onehot', OneHotEncoder(drop='first'), [3])], remainder='passthrough')
```

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)
X_train = ct.fit_transform(X_train)
```

```python
X_test = ct.transform(X_test)
```

```python
X_test
```

```python
# SVR -- Support Vector Regression

from sklearn.svm import SVR

svr = SVR()

```

```python
svr.fit(X_train, y_train)
```

```python
y_pred = svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

print(f'MAE: {mean_absolute_error(y_test, y_pred)}')
print(f'R2 Score: {r2_score(y_test, y_pred)}')
```

```python
## Hyperparameter Tuning Using GridSearchCV

from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'C': [1, 10, 100, 1000, 10000],
    'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
    'kernel' : ['linear', 'rbf', 'sigmoid'],
    'epsilon' : [0.001, 0.01, 0.1, 1,],
}
```

```python
grid = GridSearchCV(SVR(), param_grid=param_grid, refit=True, verbose=3, cv=3)

grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
svr = SVR(C=10000, epsilon= 1, gamma= 0.0001, kernel ='rbf').fit(X_train, y_train)
```

```python
y_pred = svr.predict(X_test)
print(f'MAE: {mean_absolute_error(y_test, y_pred)}')
print(f'R2 Score: {r2_score(y_test, y_pred)}')
```
