# Support Vector Classifier Implimentaion

```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
## Lets create synthetic data points

from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=1000, n_features=2, n_classes=2, n_clusters_per_class=2, n_redundant=0)
```

```python
X
```

```python
y
```

```python
sns.scatterplot(x=pd.DataFrame(X)[0],y= pd.DataFrame(X)[1], hue=y)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state=10)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='linear')
```

```python
svc.fit(X_train, y_train)
```

```python
## Prediction

y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(f'CM\n{confusion_matrix(y_test, y_pred)}')
print(f'\n\n\nClassification Report\n\n{classification_report(y_test, y_pred)}')
```

```python
# rbf kernel
svc = SVC(kernel='rbf')
svc.fit(X_train, y_train)
y_pred = svc.predict(X_test)
print(f'CM\n{confusion_matrix(y_test, y_pred)}')
print(f'\n\n\nClassification Report\n\n{classification_report(y_test, y_pred)}')
```

```python
# polynomial kernel
svc = SVC(kernel='poly')
svc.fit(X_train, y_train)
y_pred = svc.predict(X_test)
print(f'CM\n{confusion_matrix(y_test, y_pred)}')
print(f'\n\n\nClassification Report\n\n{classification_report(y_test, y_pred)}')
```

```python
# sigmoid kernel
svc = SVC(kernel='sigmoid')
svc.fit(X_train, y_train)
y_pred = svc.predict(X_test)
print(f'CM\n{confusion_matrix(y_test, y_pred)}')
print(f'\n\n\nClassification Report\n\n{classification_report(y_test, y_pred)}')
```

## Hyperparameter Tuning with SVC

```python


from sklearn.model_selection import GridSearchCV

# defining parameter range
param_grid = {
    'C': [0.1, 1, 10, 100, 1000],
    'gamma' : [1, 0.1, 0.01, 0.001, 0.0001],
    'kernel' : ['rbf', 'linear', 'poly']
}


grid = GridSearchCV(SVC(), param_grid=param_grid, cv=5, refit=True, verbose=3)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
print(f'CM\n{confusion_matrix(y_test, y_pred)}')
print(f'\n\n\nClassification Report\n\n{classification_report(y_test, y_pred)}')
```
