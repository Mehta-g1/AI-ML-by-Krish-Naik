## Why data is missing

1. **Missing Completely at Random (MCAR)**
   - The missing data is unrelated to any variable in the dataset
   - It happens randomly with no pattern or reason
   - Examples:
     - A sensor randomly malfunctions and fails to record temperature readings
     - A participant accidentally skips a question on a survey
     - A database connection drops and some records aren't saved
     - A hard drive failure causes random data loss across files

2. **Missing at Random (MAR)**
   - The missing data is related to other variables we can see, but not to the missing value itself
   - The pattern of missingness depends on other observed variables
   - Examples:
     - Younger people are less likely to fill in their age field
     - Patients with less severe illnesses skip certain medical tests
     - Lower-income households are less likely to respond to income-related questions
     - Students absent on a particular day miss the test that was given that day

3. **Missing Not at Random (MNAR)**
   - The missing data is directly related to the value that's missing
   - The reason for missingness is unknown and depends on the unobserved value
   - Examples:
     - People with high incomes might not report their salary (high earners refuse to answer)
     - Depressed individuals skip depression survey questions
     - Students who score very low on tests might not submit them
     - Overweight people might skip weight measurement questions
