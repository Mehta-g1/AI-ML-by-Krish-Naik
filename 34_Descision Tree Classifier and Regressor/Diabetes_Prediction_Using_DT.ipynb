```python
from sklearn.datasets import load_diabetes
```

```python
dataset = load_diabetes()
```

```python
print(dataset)
```

```python
print(dataset['DESCR'])
```

```python
import pandas as pd

df_diabetes = pd.DataFrame(
    dataset.data,
        columns = [
        "age",
        "sex", 
        "bmi",
        "bp",
        "s1",
        "s2",
        "s3",
        "s4",
        "s5",
        "s6",
    ]
)
```

```python
df_diabetes.sample(5)
```

```python
# Independent and dependent feature
X = df_diabetes
y = dataset['target']
```

```python
from sklearn.model_selection import  train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=10, test_size=0.3)
```

```python
X_train.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize = (15, 10))
sns.heatmap(X_train.corr(), annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor

regressor = DecisionTreeRegressor()
regressor.fit(X_train, y_train)
```

## Hyperparameter Tuning

```python
param = {
    "criterion": ["squared_error", "friedman_mse" "absolute_error",],
    "splitter" : ["best", "random"],
    "max_features": ["auto", "sqrt", "log2"],
    "max_depth" : [1,2,3,4,5, 10, 15, 20, 25]

}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
regressor = DecisionTreeRegressor()
```

```python
grid = GridSearchCV(regressor, param_grid=param, cv=5, scoring='neg_mean_squared_error')
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
y_pred = grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
print("r2_score :",r2_score(y_test, y_pred))
print("mean_absolute_error : ", mean_absolute_error(y_test, y_pred))
print("mean_squared_error : ", mean_squared_error(y_test, y_pred))
```

```python
selectmodel = DecisionTreeRegressor(criterion='squared_error', max_depth=5, max_features='sqrt', splitter='best')
```

```python
selectmodel.fit(X_train, y_train)
```

```python
## Visualize the Descision Tree

import matplotlib.pyplot as plt
from sklearn import tree

plt.figure(figsize=(15, 10))

tree.plot_tree(selectmodel, filled=True)

plt.savefig('output.png')
plt.show()
```
