### 1. What is a Decision Tree?

A **Decision Tree** is a machine learning model that makes decisions much like a human does—by asking a series of questions. Think of it as a flowchart where:
*   **Root Node (The starting question):** The very first choice/split point.
*   **Decision Nodes (Sub-questions):** Mid-level conditions based on features.
*   **Leaf Nodes (Final answers):** The final prediction or outcome (e.g., *Yes, play tennis* or *No, do not play tennis*).

#### Our Goal with this Dataset:
We want to predict the target variable **`Play Tennis`** (which has values `Yes` or `No`) based on four features:
1.  **Outlook** (`Sunny`, `Overcast`, `Rain`)
2.  **Temperature** (`Hot`, `Mild`, `Cool`)
3.  **Humidity** (`High`, `Normal`)
4.  **Wind** (`Weak`, `Strong`)

---

### 2. Our Starting Point (The Parent Node)

If we look at all **14 days** in the dataset:
*   **Total Days ($N$):** 14
*   **Days we played Tennis ($Yes$):** 9 days (Days 3, 4, 5, 7, 9, 10, 11, 12, 13)
*   **Days we did not play ($No$):** 5 days (Days 1, 2, 6, 8, 14)

Our initial state has **9 Yes** and **5 No** answers. This mix is called **impure** because it's not all "Yes" or all "No". To build a decision tree, we need mathematical ways to measure this impurity. That's where **Entropy** and **Gini Impurity** come in.

---

### 3. What is Entropy? (The Measure of "Chaos")

**Entropy** measures the amount of uncertainty, randomness, or impurity in a dataset. 
*   If a node has a **50/50 split** (e.g., 7 Yes, 7 No), it is completely impure, and its Entropy is **1.0** (maximum chaos).
*   If a node contains only one class (e.g., 14 Yes, 0 No), it is perfectly pure, and its Entropy is **0.0** (zero chaos).

#### The Formula:
$$Entropy(S) = -p_{yes} \log_2(p_{yes}) - p_{no} \log_2(p_{no})$$

Where:
*   $p_{yes}$ is the probability of a "Yes" = $\frac{\text{Number of Yes}}{\text{Total Days}}$
*   $p_{no}$ is the probability of a "No" = $\frac{\text{Number of No}}{\text{Total Days}}$

#### Step-by-Step Calculation for our Parent Node:
1.  $p_{yes} = \frac{9}{14} \approx 0.643$
2.  $p_{no} = \frac{5}{14} \approx 0.357$
3.  Calculate the logs:
    *   $\log_2(0.643) \approx -0.637$
    *   $\log_2(0.357) \approx -1.485$
4.  Plug them in:
    $$Entropy(S) = - (0.643 \times -0.637) - (0.357 \times -1.485)$$
    $$Entropy(S) = 0.410 + 0.530 = 0.940$$

**Initial Entropy is 0.940.** (It is quite high because the dataset is mixed).

---

### 4. What is Gini Impurity? (The Faster Alternative)

**Gini Impurity** is another way to measure impurity. While Entropy uses logarithmic functions (which are computationally heavy), Gini Impurity uses simple squaring, making it much faster to compute.
*   Gini ranges from **0.0** (perfectly pure) to **0.5** (perfectly impure for a 50/50 binary split).

#### The Formula:
$$Gini(S) = 1 - (p_{yes}^2 + p_{no}^2)$$

#### Step-by-Step Calculation for our Parent Node:
1.  $p_{yes} = \frac{9}{14} \approx 0.643$
2.  $p_{no} = \frac{5}{14} \approx 0.357$
3.  Square them and subtract from 1:
    $$Gini(S) = 1 - (0.643^2 + 0.357^2)$$
    $$Gini(S) = 1 - (0.413 + 0.127) = 1 - 0.540 = 0.460$$

**Initial Gini Impurity is 0.460.**

---

### 5. What is Information Gain? (Choosing the Best Split)

When building a tree, we need to choose which feature (`Outlook`, `Temperature`, `Humidity`, or `Wind`) to split on first. 

**Information Gain** measures how much the entropy (or Gini impurity) decreases after we split the data by a certain feature. We always choose the feature that gives us the **highest Information Gain** (the greatest reduction in chaos).

#### The Formula:
$$Information\ Gain = Entropy(\text{Parent}) - [\text{Weighted Average of Entropies of Children}]$$

Let's compare splitting on two features: **Wind** vs. **Outlook**.

---

#### Scenario A: Splitting on `Wind`
`Wind` has two branches: **Weak** and **Strong**.

1.  **Branch 1: Wind is Weak (8 Days total)**
    *   Days: 1, 3, 4, 5, 8, 9, 10, 13
    *   Play Tennis outcomes: **6 Yes, 2 No**
    *   $Entropy(S_{weak}) = -(\frac{6}{8}\log_2\frac{6}{8} + \frac{2}{8}\log_2\frac{2}{8}) \approx 0.811$

2.  **Branch 2: Wind is Strong (6 Days total)**
    *   Days: 2, 6, 7, 11, 12, 14
    *   Play Tennis outcomes: **3 Yes, 3 No**
    *   $Entropy(S_{strong}) = -(\frac{3}{6}\log_2\frac{3}{6} + \frac{3}{6}\log_2\frac{3}{6}) = 1.0$ (Perfect 50/50 split)

3.  **Weighted Entropy after split:**
    $$Weighted\ Entropy = \left(\frac{8}{14} \times 0.811\right) + \left(\frac{6}{14} \times 1.0\right) = 0.463 + 0.429 = 0.892$$

4.  **Information Gain for Wind:**
    $$IG(Wind) = 0.940 - 0.892 = 0.048$$

---

#### Scenario B: Splitting on `Outlook`
`Outlook` has three branches: **Sunny**, **Overcast**, and **Rain**.

1.  **Branch 1: Outlook is Sunny (5 Days total)**
    *   Play Tennis outcomes: **2 Yes, 3 No** (Days 1, 2, 8, 9, 11)
    *   $Entropy(S_{sunny}) = -(\frac{2}{5}\log_2\frac{2}{5} + \frac{3}{5}\log_2\frac{3}{5}) \approx 0.971$

2.  **Branch 2: Outlook is Overcast (4 Days total)**
    *   Play Tennis outcomes: **4 Yes, 0 No** (Days 3, 7, 12, 13)
    *   $Entropy(S_{overcast}) = 0.0$ (Perfectly pure! If it is overcast, we *always* play tennis).

3.  **Branch 3: Outlook is Rain (5 Days total)**
    *   Play Tennis outcomes: **3 Yes, 2 No** (Days 4, 5, 6, 10, 14)
    *   $Entropy(S_{rain}) = -(\frac{3}{5}\log_2\frac{3}{5} + \frac{2}{5}\log_2\frac{2}{5}) \approx 0.971$

4.  **Weighted Entropy after split:**
    $$Weighted\ Entropy = \left(\frac{5}{14} \times 0.971\right) + \left(\frac{4}{14} \times 0.0\right) + \left(\frac{5}{14} \times 0.971\right) = 0.347 + 0.0 + 0.347 = 0.694$$

5.  **Information Gain for Outlook:**
    $$IG(Outlook) = 0.940 - 0.694 = 0.246$$

---

### The Decision:
*   **Information Gain (Wind) = 0.048**
*   **Information Gain (Outlook) = 0.246**

Because **Outlook** yields a much higher Information Gain (0.246 vs 0.048), the Decision Tree algorithm will select **Outlook** as the very first split (the Root Node) of our tree.

```
                  [Outlook?]
                 /    |     \
         (Sunny)  (Overcast)  (Rain)
          /           |          \
    [Humidity?]    [ YES ]     [Wind?]
     (Pure Class)
```

---

### Summary Checklist:
1.  **Decision Tree:** A structure of consecutive yes/no questions designed to classify or predict data.
2.  **Entropy:** A metric measuring how mixed/uncertain the data is ($0$ = perfectly pure, $1$ = completely mixed).
3.  **Gini Impurity:** A mathematically simpler and faster alternative to Entropy ($0$ = perfectly pure, $0.5$ = completely mixed).
4.  **Information Gain:** The difference in impurity before and after splitting. The feature with the highest Information Gain is chosen for the split.

### 6. Decision Tree Split for Numerical Features

When dealing with **categorical features** (like *Outlook* or *Wind*), splitting is straightforward because the feature has distinct, predefined categories. However, for **numerical features** (like temperature values, height, salary, etc.), the data is continuous. We need a way to convert these continuous values into binary split decisions (e.g., $f_1 \leq \text{Threshold}$ vs $f_1 > \text{Threshold}$).

Here is the step-by-step algorithm used by Decision Trees to split on a numerical feature:

---

#### Step 1: Sort the Feature Values
First, sort the numerical feature values in ascending order along with their corresponding target output ($O/p$).

Using the example from the reference document [2_Decision+Tree+split+For+Numerical+Features.pdf](file:///g:/My%20Drive/AL-ML%20by%20Krish%20Naik/34_Descision%20Tree%20Classifier%20and%20Regressor/2_Decision+Tree+split+For+Numerical+Features.pdf):
* **Sorted Feature ($f_1$):** `2.3`, `3.6`, `4.0`, `5.2`, `6.7`, `8.9`, `10.5`
* **Output / Target:**
  | Feature Value ($f_1$) | Target ($O/p$) |
  |-----------------------|----------------|
  | 2.3                   | Yes            |
  | 3.6                   | Yes            |
  | 4.0                   | No             |
  | 5.2                   | No             |
  | 6.7                   | Yes            |
  | 8.9                   | No             |
  | 10.5                  | Yes            |

* **Parent Node Distribution:** 4 Yes, 3 No ($N_{total} = 7$)
  * $Entropy(\text{Parent}) \approx 0.985$

---

#### Step 2: Define Candidate Thresholds
We need to test different candidate thresholds to see which one splits the data best. There are two common approaches to define candidate thresholds:
1. **Using the values themselves:** Candidate thresholds are $\{2.3, 3.6, 4.0, 5.2, 6.7, 8.9, 10.5\}$.
2. **Using midpoints between consecutive values:** Candidate thresholds are the average of adjacent values: $\{\frac{2.3+3.6}{2}, \frac{3.6+4.0}{2}, \dots\} = \{2.95, 3.8, 4.6, 5.95, 7.8, 9.7\}$.

---

#### Step 3: Evaluate Information Gain for Each Candidate Threshold
For each candidate threshold, split the data into two branches (Left: $f_1 \le \text{Threshold}$, Right: $f_1 > \text{Threshold}$) and compute the Information Gain.

Let's evaluate the first two candidate thresholds (using the values themselves as in the PDF) using Entropy:

##### **Candidate 1: Threshold = 2.3**
* **Left Branch ($f_1 \leq 2.3$):** Contains $\{2.3\}$ $\rightarrow$ **1 Yes, 0 No**
  * $Entropy(\text{Left}) = 0.0$ (pure)
* **Right Branch ($f_1 > 2.3$):** Contains $\{3.6, 4.0, 5.2, 6.7, 8.9, 10.5\}$ $\rightarrow$ **3 Yes, 3 No**
  * $Entropy(\text{Right}) = 1.0$ (totally impure)
* **Weighted Entropy:** $\left(\frac{1}{7} \times 0.0\right) + \left(\frac{6}{7} \times 1.0\right) \approx 0.857$
* **Information Gain:** $IG(2.3) = 0.985 - 0.857 \approx 0.128$

##### **Candidate 2: Threshold = 3.6**
* **Left Branch ($f_1 \leq 3.6$):** Contains $\{2.3, 3.6\}$ $\rightarrow$ **2 Yes, 0 No**
  * $Entropy(\text{Left}) = 0.0$ (pure)
* **Right Branch ($f_1 > 3.6$):** Contains $\{4.0, 5.2, 6.7, 8.9, 10.5\}$ $\rightarrow$ **2 Yes, 3 No**
  * $Entropy(\text{Right}) = - (\frac{2}{5} \log_2 \frac{2}{5} + \frac{3}{5} \log_2 \frac{3}{5}) \approx 0.971$
* **Weighted Entropy:** $\left(\frac{2}{7} \times 0.0\right) + \left(\frac{5}{7} \times 0.971\right) \approx 0.694$
* **Information Gain:** $IG(3.6) = 0.985 - 0.694 \approx 0.291$

**Comparison:** Since $IG(3.6) > IG(2.3)$, splitting at $3.6$ gives higher Information Gain and is a better split point (as marked with a checkmark in the reference drawing).

```
          [ Threshold = 2.3 ]                        [ Threshold = 3.6 ]
             /           \                              /           \
     (<=2.3) /           \ (>2.3)               (<=3.6) /           \ (>3.6)
            /             \                            /             \
       [1 Y, 0 N]     [3 Y, 3 N]                  [2 Y, 0 N]     [2 Y, 3 N]
       Pure           Impure                      Pure           Impure
       IG = 0.128                                 IG = 0.291 (Higher!)
```

---

#### The Computational Bottleneck (Time Complexity)
* **Problem:** If a dataset has $N$ samples, sorting takes $\mathcal{O}(N \log N)$ time.
* **Evaluation:** For $N$ unique numerical values, there are $N-1$ candidate thresholds. We must evaluate Information Gain for each of these candidates.
* **Scale:** For **millions of records**, evaluating every threshold becomes extremely computationally expensive (Time Complexity $\uparrow\uparrow$).
* **Optimization:** In practice, modern libraries like XGBoost or LightGBM use **Histogram-based techniques** or **approximate split finding** (binning continuous values into a fixed number of bins, e.g., 256 bins) to speed up training.
